```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

#load dataset Iris dari scikit-learn
iris = load_iris()
```

- numpy digunakan untuk operasi array.
- pandas digunakan untuk mengelola data dalam bentuk tabel (DataFrame).
- load_iris digunakan untuk memuat dataset Iris dari scikit-learn.

```python
#konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]

#definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]

print(X)
```

- Data Iris dikonversi menjadi DataFrame dan ditambahkan kolom Species sebagai nama spesies.
- X adalah fitur (sepal length, sepal width, petal length, petal width).
- y adalah target/label (spesies bunga).

```python
y
```

Menampilkan isi y, yaitu label spesies dalam bentuk teks (setosa, versicolor, virginica).

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

- Label berupa teks perlu diubah menjadi angka agar dapat diproses model.
- LabelEncoder mengubah setosa menjadi 0, versicolor menjadi 1, dan virginica menjadi 2.

```python
from sklearn.model_selection import train_test_split

X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

- Data dibagi menjadi data latih (80%) dan data uji (20%).
- test_size=0.2 menentukan porsi data uji.
- random_state=0 membuat hasil pembagian data selalu sama setiap kali dijalankan.

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

- GaussianNB() membuat model Gaussian Naïve Bayes.
- fit() melatih model menggunakan data latih (X_train dan y_train).

```python
result = NB_model.predict(X_test)
result
```

- predict() digunakan untuk memprediksi kelas dari data uji (X_test).
- Hasilnya berupa angka 0, 1, atau 2 sesuai hasil encoding.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

- accuracy_score membandingkan hasil prediksi dengan label sebenarnya pada data uji.
- Akurasi 96,67% berarti 29 dari 30 data uji diprediksi dengan benar.

```python
#prediksi inputan data baru
#input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

- Input baru berupa list [[5.1, 3.5, 1.4, 0.2]] diprediksi menggunakan model yang sudah dilatih.
- inverse_transform mengubah hasil prediksi berupa angka kembali menjadi nama spesies.
- Ketika GaussianNB dilatih (fit()) dengan dataset Iris, fitur-fitur dalam dataset memiliki nama kolom (iris.feature_names).
- Namun, saat melakukan prediksi dengan list/NumPy array langsung, fitur tidak memiliki nama, sehingga muncul peringatan.

```python
input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)

result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

Input diubah menjadi DataFrame dengan nama kolom iris.feature_names, sehingga strukturnya konsisten dengan data latih dan tidak ada peringatan.
